# Literature acquisition checks
Metadata candidates are not read papers or independent experiments. Run from the project root or this notebook directory.

```python
from pathlib import Path
import json, hashlib
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts/collect_literature.py').exists())
folder = root / 'storage/research/literature-2026-09-27'
catalog = json.loads((folder/'catalog.json').read_text(encoding='utf-8'))
receipt = json.loads((folder/'receipt.json').read_text(encoding='utf-8'))
assert len(catalog) == len({r['doi'] for r in catalog})
assert all(r['training_allowed'] is False for r in catalog)
for q in receipt['queries']:
    assert hashlib.sha256((root/q['raw_path']).read_bytes()).hexdigest() == q['sha256']
receipt['statistics']
```

```python
summary = json.loads((folder/'screening-summary.json').read_text(encoding='utf-8'))
assert sum(summary['buckets'].values()) == len(catalog)
summary
```
